```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
```

```python
#load dataset
df=pd.read_csv('car data.csv')
df.head()
```

```python
#drop car_name column
df=df.drop(columns=['Car_Name'])
```

```python
#convert categorical into numerical
df_encoded=pd.get_dummies(df,drop_first=True)
df_encoded.head()
```

```python
x=df_encoded.drop('Selling_Price',axis=1)
y=df_encoded['Selling_Price']
```

```python
#train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
#train ml model random forest
from sklearn.ensemble import RandomForestRegressor
model=RandomForestRegressor(
      n_estimators=200,
      criterion="squared_error",
      random_state=42
)
model_fit=model.fit(x_train,y_train)
```

```python
#evaluate model
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error
y_pred=model_fit.predict(x_test)

print("Mean Squared Error:",mean_squared_error(y_test,y_pred))
print("Mean Absolute error:",mean_absolute_error(y_test,y_pred))
print("R2 Score:",r2_score(y_test,y_pred))
```

```python
def predict_car_price(year,
                      present_price,
                      km_driven,
                      owner,
                      company,
                      fuel_type,
                      seller_type,
                      transmission):
  # Create a DataFrame with the input features, using original column names as they were in the raw data
  input_df_dict = {
      "Year": [year],
      "Present_Price": [present_price],
      "Kms_Driven": [km_driven],
      "Owner": [owner],
      "company": [company],
      "Fuel_Type": [fuel_type],
      "Seller_Type": [seller_type],
      "Transmission": [transmission]
  }

  input_data = pd.DataFrame(input_df_dict)

  # Convert categorical columns to numerical using one-hot encoding, consistent with df_encoded
  input_data_encoded = pd.get_dummies(input_data, drop_first=True)

  # Reindex the encoded input to match the columns of the training data (x)
  # Fill missing columns with 0 and drop any extra columns
  input_data_reindexed = input_data_encoded.reindex(columns=x.columns, fill_value=0)

  # Make the prediction
  prediction = model_fit.predict(input_data_reindexed)[0]

  return prediction # Return the prediction
```

```python
price=predict_car_price(
    year=2019,
    present_price=9.0,
    km_driven=500000,
    owner=0,
    company="Maruti",
    fuel_type="Petrol",
    seller_type="Individual",
    transmission="Manual"
)
print("predicted selling price",price)
```
